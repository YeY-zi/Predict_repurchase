# 第2课：90天不是标准答案——怎样选择预测窗口？

## 决策顺序
业务行动与等待成本 → 数据覆盖 → 候选窗口 → 样本量与稳定性 → 明确取舍并冻结定义。

问题不是“哪个窗口能让模型最好看”，而是“哪个期限对应一个有用、能可靠评价的问题”。
30、60、90、120、180天只是候选值，不是数据自动发现的结论。
历史窗口与预测窗口是两个独立选择；本课暂固定六个月历史，以单独研究预测期限。

本课为探索分析，不是独立最终测试。尤其2018-04-01快照及其结果已经看过，不能再称为未触碰测试。
仍用已批准且整单商品金额为正的订单；不依据后来最终状态过滤历史样本。
数据来自你们仓库，原包许可 CC BY-NC-SA 4.0。


第一步：先问“预测结果准备用来做什么”，暂时不定90天。

例如，你独自拿到 Olist 数据，发现它有买家标识、订单日期和金额，可以提出一个候选问题：
根据历史购买行为，能否找出将来更可能回来购买的客户？

接着必须问“将来多久”：
| 假设的业务需求 | 值得研究的候选期限 |
|---|---|
| 希望一个月内看到结果 | 30天 |
| 允许观察两三个月 | 60天、90天 |
| 关注更长期的客户回访 | 120天、180天 |
    第二步：检查数据能不能支撑这些期限。

比如选择2018年4月1日作为预测时点：
    预测未来30天  → 需要观察至2018年5月1日，不含当天
预测未来90天  → 需要观察至2018年6月30日，不含当天
预测未来180天 → 需要观察至2018年9月28日，不含当天

>2018年7月：6,164笔
>2018年8月：6,607笔
>2018年9月：    1笔

这让我们对9月覆盖产生疑问。因此，练习中暂时把 2018年9月1日作为不含当天的观察上界，不依赖9月记录判断完整结果。这是一项保守的工作假设，仍不能单独证明之前的数据完全无遗漏。

相当于目前9月只有1笔，那9月就没有必要作为验证集了吧，因为我们要监督预测，9月根本没数据啊？？？？？

第三步：固定一批客户，比较不同窗口的标签数量。

为了只研究未来期限，我们暂时固定：
- 预测日：2018年1月1日；
- 客户范围：此前六个月至少有一笔合格订单的客户；
- 合格订单：已批准、商品金额合计为正；
- 历史六个月也是暂用设定，之后可以单独论证。
这批客户共有 29,368人。真实计算结果是：
| 未来期限 | 观察到的复购人数 | 复购比例 |
|---|---:|---:|
| 30天 | 99 | 0.34% |
| 60天 | 164 | 0.56% |
| 90天 | 230 | 0.78% |
| 120天 | 284 | 0.97% |
| 180天 | 401 | 1.37% |
你现在可以开始作判断：
- 30天：结果反馈快，但只有99个正样本，建模和评价的信息比较有限。
- 90天：观察到230个正样本，比30天多，等待时间也更长。
- 180天：正样本更多，但需要更长观察期，也限制了可以安排的预测时点。
这里没有证据证明“230个正样本就一定够用”。 它只是告诉我们，90天具备继续做探索实验的基础；仍需考虑模型复杂度、评价不确定性和验证设计。

还有一个特别重要的判断：
窗口越长，复购人数自然可能越多。这不代表模型预测得更准。

同一客户第70天才再次购买，30天标签是0，90天标签是1。我们改变了问题的答案定义，而不是让同一道题更容易答对。

第四步：换几个预测时点，检查结论是否过度依赖某一期。

我也计算了90天窗口：
| 预测时点 | 符合条件的客户数 | 未来90天复购人数 |
|---|---:|---:|
| 2017年10月1日 | 21,160 | 211 |
| 2018年1月1日 | 29,368 | 230 |
| 2018年4月1日 | 37,813 | 299 |

本研究假设运营人员关注约一个季度内的客户回访。比较30—180天候选窗口后，发现90天在所选历史时点具有可观察的结果窗口，且比30天包含更多复购事件，因此将其作为初始研究期限。后续比较60天和120天的敏感性，不预先认定90天最优。

提出业务用途
    ↓
列出候选期限
    ↓
检查各预测时点的后续覆盖
    ↓
固定客户名单，计算各期限的复购人数和比例
    ↓
换时点检查敏感性
    ↓
权衡业务时效、样本信息和验证空间
    ↓
确定工作版期限，记录理由，再开始建模

```python
from pathlib import Path
from zipfile import ZipFile
import pandas as pd

DATA = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'data/olist-v2.zip').is_file()), Path.cwd()) / 'data' / 'olist-v2.zip'
assert DATA.exists(), '请切换到 olist_learning 文件夹运行'
def read(name):
    with ZipFile(DATA) as z:
        with z.open(name) as f:
            return pd.read_csv(f)

orders = read('olist_orders_dataset.csv')
items = read('olist_order_items_dataset.csv')
customers = read('olist_customers_dataset.csv')
orders['approved_at'] = pd.to_datetime(orders['order_approved_at'])
# 先汇总到订单粒度，避免将商品明细行误当成订单。
amounts = items.groupby('order_id')['price'].sum().rename('amount')
orders = orders.merge(amounts, on='order_id', how='left', validate='one_to_one')
orders = orders.merge(customers[['customer_id', 'customer_unique_id']],
                      on='customer_id', validate='one_to_one')
eligible_orders = orders.loc[
    orders['approved_at'].notna() & orders['amount'].gt(0)
].copy()
print('合格订单数：', len(eligible_orders))
```

## 1. 数据覆盖：最大日期不是完整覆盖的证明
先按批准月份检查。后续若能取得数据发布说明，应与其交叉确认。
本课暂以2018-09-01为不含当天的观察上界，即不使用9月及以后来支持完整窗口。
这是针对尾部稀疏现象的工作假设，不是对8月及之前完整性的证明。


```python
monthly = eligible_orders.groupby(eligible_orders['approved_at'].dt.to_period('M')).size()
print(monthly.tail(10).to_string())
OBSERVATION_END = pd.Timestamp('2018-09-01')
print('暂定观察上界（不含当天）：', OBSERVATION_END.date())
```

暂定观察上界（不含当天）： 2018-09-01这些值是不是说明了每个月有这些订单数，为什么不从原始订单购买时间覆盖2016年9月至2018年10月都看呢？## 2. 固定客户名单，只改变未来窗口
开发探索先用2018-01-01快照。名单是过去六个月至少有一笔合格订单的买家。
未来窗口不同，名单不变；标签可能改变。因此，窗口比较同时也是不同任务的比较。
现在的意思是看2018年1月1之前的6个月，然后我们用6个月的记录预测不同30, 60, 90, 120, 180未来天数的复购

## 2. 固定客户名单，只改变未来窗口
开发探索先用2018-01-01快照。名单是过去六个月至少有一笔合格订单的买家。
未来窗口不同，名单不变；标签可能改变。因此，窗口比较同时也是不同任务的比较。


```python
def compare_horizons(snapshot, horizons=(30, 60, 90, 120, 180)):
    cut = pd.Timestamp(snapshot)
    start = cut - pd.DateOffset(months=6)
    historical = eligible_orders.loc[
        eligible_orders['approved_at'].ge(start)
        & eligible_orders['approved_at'].lt(cut)
    ]
    ids = pd.Index(historical['customer_unique_id'].unique())
    rows = []
    for horizon in horizons:
        end = cut + pd.Timedelta(days=horizon)
        covered = end <= OBSERVATION_END
        if covered:
            future = eligible_orders.loc[
                eligible_orders['approved_at'].ge(cut)
                & eligible_orders['approved_at'].lt(end), 'customer_unique_id'
            ]
            count = int(ids.isin(future).sum())
        else:
            count = None  # 不为观察不足的窗口输出貌似完整的复购率
        rows.append({'snapshot': str(cut.date()), 'horizon_days': horizon,
                     'customers': len(ids), 'end_exclusive': str(end.date()),
                     'within_working_boundary': covered, 'repeaters': count,
                     'repeat_rate': count / len(ids) if count is not None and len(ids) else None})
    return pd.DataFrame(rows)

jan = compare_horizons('2018-01-01')
print(jan.to_string(index=False))
```

## 3. 换时点看敏感性
2017-10-01与2018-04-01也检查一次。不同快照可能有重复客户，不是独立重复试验。
本格全部属于探索；不能根据已观察的结果反复选窗口后，仍把相同时期叫最终独立测试。


```python
for date in ['2017-10-01', '2018-04-01']:
    print(compare_horizons(date).to_string(index=False))
    print()
```

## 4. 怎样写结论？
可以说：在当前订单定义和历史窗口下，90天在若干快照中可以观察到约200—300名复购者，具备继续开展探索比较的基础；事件仍非常稀少，需要谨慎验证。数字不构成统计充分性保证。

不能说：90天是最优；180天正样本多所以一定好；预测90天分数比30天高所以算法更强。

如果业务每月行动并要求短期回报，30天可能更贴切；若业务允许等待一个季度，90天可能是折中候选。没有真实业务方时，应明确这是课程研究的场景假设。

还要考虑：更长预测窗口需要等待更久才能获得标签，并减少可安排的时间验证时点。
如果根据验证结果选择任务与模型，应保留新的时间评估方案；不够数据时就承认探索性质。

## 为什么不直接看“已复购客户的平均间隔”？
那只包含已经观察到再次购买的人，忽略没有再次购买或观察不足者。
相邻订单也可能在同一天，需理解是否属于业务上想研究的复购。
购买间隔可用来提出候选窗口，但不能单独决定预测期限。需要估计回购时间时，可进一步学习删失与生存分析。
